```python
# =====================================================================
#  Trustworthy OSCC detection with pathology foundation models  (v2)
#  Single-file Kaggle script (paste into one notebook cell, or run as .py)
#
#  Changes vs v1
#   - PLIP fix for new transformers (calls vision tower + projection directly)
#   - every model wrapped in try/except: one failure never kills the run
#   - images cached at 448x672 and features are the MEAN over six 224x224
#     tiles (keeps nuclear detail; whole-image 224 squash was the weak spot)
#   - fine-tuned baselines: ImageNet ViT-B/16 AND full fine-tune of Phikon
#   - prints sample filenames per split (check for augmented copies)
#
#  Kaggle settings: Accelerator = GPU, Internet = ON.
#  UNI needs a Kaggle Secret HF_TOKEN + accepted licence; otherwise skipped.
# =====================================================================

# %% ------------------------------ 0. SETUP ---------------------------
import os, sys, glob, json, math, time, random, warnings, subprocess
warnings.filterwarnings("ignore")

def _pip(*pkgs):
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", *pkgs], check=False)

for _mod, _pkg in [("timm", "timm"), ("transformers", "transformers"),
                   ("sklearn", "scikit-learn")]:
    try:
        __import__(_mod)
    except ImportError:
        _pip(_pkg)

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import roc_auc_score, confusion_matrix
from sklearn.model_selection import StratifiedKFold, GroupKFold
from sklearn.cluster import AgglomerativeClustering
import timm
from torchvision import transforms as T

CFG = dict(
    DATA_ROOT=None,                 # None = auto-find under /kaggle/input
    IMG=224, HI_H=448, HI_W=672,    # cached res = 2x3 grid of 224 tiles
    BATCH=64, SEED=42, NUM_WORKERS=2,
    MODELS=["phikon", "phikon_v2", "uni", "dinov2", "plip"],
    CONCEPT_MODEL="plip",           # must be in MODELS
    GROUP_MODEL="dinov2",           # D4-averaged whole-image embedding for leakage audit
    FINETUNE=["vit_b16", "phikon"], # [] to skip
    FT_EPOCHS=5, FT_LR=3e-5, FT_BS=32,
    ALPHA=0.05,                     # conformal miscoverage (0.05 -> 95% coverage)
    GROUP_DIST_THRESH=0.15,
    CV_FOLDS=5,
    OUT="/kaggle/working",
)
BASE = dict(acc=97.69, prec=98.21, sens=97.51)   # paper ViT-14 (Table 21)

def seed_all(s):
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
    torch.cuda.manual_seed_all(s)

seed_all(CFG["SEED"])
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
AMP = DEVICE == "cuda"
print(f"device={DEVICE}  torch={torch.__version__}  timm={timm.__version__}")
os.makedirs(CFG["OUT"], exist_ok=True)

# %% ------------------------- 1. DATA DISCOVERY -----------------------
def find_root():
    if CFG["DATA_ROOT"]:
        return CFG["DATA_ROOT"]
    for dp, dn, _ in os.walk("/kaggle/input"):
        if {"train", "val", "test"} <= {d.lower() for d in dn}:
            return dp
    raise FileNotFoundError("Could not find train/val/test folders under /kaggle/input")

ROOT = find_root()
print("data root:", ROOT)
EXT = (".jpg", ".jpeg", ".png", ".tif", ".tiff", ".bmp")

def collect(root):
    paths, labels, splits = [], [], []
    for sp in os.listdir(root):
        if sp.lower() not in ("train", "val", "test"):
            continue
        for cls in os.listdir(os.path.join(root, sp)):
            lab = {"normal": 0, "oscc": 1}.get(cls.lower())
            if lab is None:
                continue
            for f in sorted(os.listdir(os.path.join(root, sp, cls))):
                if f.lower().endswith(EXT):
                    paths.append(os.path.join(root, sp, cls, f))
                    labels.append(lab); splits.append(sp.lower())
    return paths, np.array(labels), np.array(splits)

PATHS, Y, SPLIT = collect(ROOT)
print(pd.crosstab(SPLIT, Y, rownames=["split"], colnames=["0=Normal 1=OSCC"]))
for sp in ("train", "val", "test"):
    names = [os.path.basename(p) for p, s in zip(PATHS, SPLIT) if s == sp]
    print(f"  {sp} sample files: {names[:3]} ... {names[-3:]}")
TR, VA, TE = (np.where(SPLIT == s)[0] for s in ("train", "val", "test"))
N = len(PATHS)
TILE, HH, WW = CFG["IMG"], CFG["HI_H"], CFG["HI_W"]
GR, GC = HH // TILE, WW // TILE
NT = GR * GC

class ImgDS(torch.utils.data.Dataset):
    def __len__(self): return N
    def __getitem__(self, i):
        im = Image.open(PATHS[i])
        try: im.draft("RGB", (WW * 2, HH * 2))
        except Exception: pass
        im = im.convert("RGB").resize((WW, HH), Image.BICUBIC)
        return torch.from_numpy(np.asarray(im).copy()).permute(2, 0, 1)

t0 = time.time()
X_hi = torch.empty((N, 3, HH, WW), dtype=torch.uint8)
_i = 0
for b in torch.utils.data.DataLoader(ImgDS(), batch_size=32, num_workers=CFG["NUM_WORKERS"]):
    X_hi[_i:_i + len(b)] = b; _i += len(b)
# whole-image 224x224 view (used by fine-tuning and the leakage audit)
X224 = torch.cat([F.interpolate(X_hi[i:i + 256].float(), size=(TILE, TILE), mode="bilinear",
                                antialias=True, align_corners=False).round().clamp(0, 255).to(torch.uint8)
                  for i in range(0, N, 256)])
print(f"cached hi-res {tuple(X_hi.shape)} + whole {tuple(X224.shape)} in {time.time()-t0:.0f}s "
      f"({X_hi.numel()/1e9:.1f} GB)")

_cj = T.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.4, hue=0.08)
def perturb(xb):
    """colour/stain-style perturbation on uint8 tensors, same for every model"""
    g = torch.get_rng_state(); torch.manual_seed(1234)
    out = torch.stack([_cj(im) for im in xb])
    torch.set_rng_state(g)
    return out

# %% --------------------------- 2. METRICS ----------------------------
def ece_score(p, y, bins=15):
    conf, pred = p.max(1), p.argmax(1)
    hit = (pred == y).astype(float); e = 0.0
    edges = np.linspace(0, 1, bins + 1)
    for lo, hi in zip(edges[:-1], edges[1:]):
        m = (conf > lo) & (conf <= hi)
        if m.any(): e += m.mean() * abs(hit[m].mean() - conf[m].mean())
    return e * 100

def clf_metrics(y, p):
    pred = p.argmax(1)
    tn, fp, fn, tp = confusion_matrix(y, pred, labels=[0, 1]).ravel()
    acc = (tp + tn) / len(y); prec = tp / max(tp + fp, 1)
    sens = tp / max(tp + fn, 1); spec = tn / max(tn + fp, 1)
    f1 = 2 * prec * sens / max(prec + sens, 1e-9)
    return dict(acc=float(acc * 100), prec=float(prec * 100), sens=float(sens * 100),
                spec=float(spec * 100), f1=float(f1 * 100),
                auc=float(roc_auc_score(y, p[:, 1]) * 100), ece=float(ece_score(p, y)))

def boot_ci(y, pred, n=1000, seed=0):
    rng = np.random.default_rng(seed); a = []
    for _ in range(n):
        i = rng.integers(0, len(y), len(y)); a.append((y[i] == pred[i]).mean())
    return np.percentile(a, [2.5, 97.5]) * 100

def conformal(p_cal, y_cal, p_te, y_te, alpha):
    n = len(y_cal)
    s = 1 - p_cal[np.arange(n), y_cal]
    q = np.quantile(s, min(1.0, math.ceil((n + 1) * (1 - alpha)) / n), method="higher")
    sets = (1 - p_te) <= q
    sz = sets.sum(1); single = sz == 1
    return dict(coverage=float(sets[np.arange(len(y_te)), y_te].mean() * 100),
                avg_set_size=float(sz.mean()), frac_singleton=float(single.mean() * 100),
                frac_abstain_both=float((sz == 2).mean() * 100),
                acc_on_singleton=(float((sets[single].argmax(1) == y_te[single]).mean() * 100)
                                  if single.any() else float("nan")))

def add_row(RES, label, p_va, p_te, p_tp):
    m = clf_metrics(Y[TE], p_te); lo, hi = boot_ci(Y[TE], p_te.argmax(1))
    cf = conformal(p_va, Y[VA], p_te, Y[TE], CFG["ALPHA"])
    RES.append(dict(model=label, **m, ci_lo=float(lo), ci_hi=float(hi),
                    acc_stain=clf_metrics(Y[TE], p_tp)["acc"],
                    **{f"cp_{k}": v for k, v in cf.items()}))
    print("  ", {k: round(v, 2) for k, v in m.items()}, f"CI95=[{lo:.1f},{hi:.1f}]",
          f"| stain acc={RES[-1]['acc_stain']:.2f}")
    print("   conformal:", {k: round(v, 2) for k, v in cf.items()})
    return cf

# %% --------------------- 3. FEATURE EXTRACTORS -----------------------
IMN = ((0.485, 0.456, 0.406), (0.229, 0.224, 0.225))
CLIPN = ((0.4815, 0.4578, 0.4082), (0.2686, 0.2613, 0.2758))
PLIP = {}

def hf_token():
    try:
        from kaggle_secrets import UserSecretsClient
        return UserSecretsClient().get_secret("HF_TOKEN")
    except Exception:
        return os.environ.get("HF_TOKEN")

def load_extractor(name):
    from transformers import AutoModel
    if name in ("phikon", "phikon_v2"):
        repo = {"phikon": "owkin/phikon", "phikon_v2": "owkin/phikon-v2"}[name]
        m = AutoModel.from_pretrained(repo).to(DEVICE).eval()
        return (lambda x: m(pixel_values=x).last_hidden_state[:, 0]), IMN
    if name == "dinov2":
        m = AutoModel.from_pretrained("facebook/dinov2-base").to(DEVICE).eval()
        def fn(x):
            h = m(pixel_values=x).last_hidden_state
            return torch.cat([h[:, 0], h[:, 1:].mean(1)], 1)
        return fn, IMN
    if name == "uni":
        tok = hf_token()
        if tok:
            from huggingface_hub import login; login(tok)
        m = timm.create_model("hf-hub:MahmoodLab/uni", pretrained=True,
                              init_values=1e-5, dynamic_img_size=True).to(DEVICE).eval()
        return (lambda x: m(x)), IMN
    if name == "plip":
        from transformers import CLIPModel, CLIPProcessor
        m = CLIPModel.from_pretrained("vinid/plip").to(DEVICE).eval()
        PLIP["model"] = m
        PLIP["proc"] = CLIPProcessor.from_pretrained("vinid/plip")
        # version-proof: call the vision tower + projection directly
        return (lambda x: m.visual_projection(m.vision_model(pixel_values=x).pooler_output)), CLIPN
    raise ValueError(name)

def _tiles(xb):
    B = xb.shape[0]
    t = [xb[:, :, r * TILE:(r + 1) * TILE, c * TILE:(c + 1) * TILE]
         for r in range(GR) for c in range(GC)]
    return torch.stack(t, 1).reshape(B * NT, 3, TILE, TILE)

@torch.no_grad()
def extract(fn, Xu8, norm, perturb_fn=None, d4=False, tiled=True):
    mean = torch.tensor(norm[0], device=DEVICE).view(1, 3, 1, 1)
    std = torch.tensor(norm[1], device=DEVICE).view(1, 3, 1, 1)
    bs = max(1, CFG["BATCH"] // NT) if tiled else CFG["BATCH"]
    out = []
    for i in range(0, len(Xu8), bs):
        xb = Xu8[i:i + bs]
        if perturb_fn is not None: xb = perturb_fn(xb)
        xb = (xb.to(DEVICE).float() / 255 - mean) / std
        B = xb.shape[0]
        if tiled: xb = _tiles(xb)
        with torch.autocast("cuda", dtype=torch.float16, enabled=AMP):
            if d4:   # average over the 8 dihedral transforms
                fs = []
                for fl in (False, True):
                    xf = xb.flip(3) if fl else xb
                    for r in range(4):
                        fs.append(fn(torch.rot90(xf, r, (2, 3))).float())
                f = torch.stack(fs).mean(0)
            else:
                f = fn(xb).float()
        if tiled: f = f.view(B, NT, -1).mean(1)
        out.append(f.cpu())
    return torch.cat(out).numpy()

def fit_probe(Ftr, ytr):
    return make_pipeline(StandardScaler(),
                         LogisticRegressionCV(Cs=[1e-3, 1e-2, 1e-1, 1, 10], cv=3,
                                              max_iter=3000, scoring="neg_log_loss")).fit(Ftr, ytr)

# %% ------------------------ 4. RUN ALL MODELS ------------------------
RES, FEATS, FEATS_P, PROBS, extra = [], {}, {}, {}, {}
F_GROUP = None

for name in CFG["MODELS"]:
    print(f"\n=== {name} (frozen, mean of {NT} tiles) ===")
    try:
        fn, norm = load_extractor(name)
        t0 = time.time()
        Fa = extract(fn, X_hi, norm)
        Fp = extract(fn, X_hi[TE], norm, perturb_fn=perturb)
        if name == CFG["GROUP_MODEL"]:
            F_GROUP = extract(fn, X224, norm, d4=True, tiled=False)
        print(f"  features {Fa.shape} in {time.time()-t0:.0f}s")
        FEATS[name], FEATS_P[name] = Fa, Fp
        clf = fit_probe(Fa[TR], Y[TR])
        p_te, p_va, p_tp = (clf.predict_proba(Fa[TE]), clf.predict_proba(Fa[VA]),
                            clf.predict_proba(Fp))
        PROBS[name] = p_te
        extra[name] = add_row(RES, f"{name} + linear probe", p_va, p_te, p_tp)
    except Exception as e:
        print(f"  skipped ({type(e).__name__}: {str(e)[:150]})")
    if name != CFG["CONCEPT_MODEL"]:
        fn = None
    torch.cuda.empty_cache()

# %% ----------------------- 5. FINE-TUNED BASELINES -------------------
class HFBackbone(nn.Module):
    def __init__(self, repo):
        super().__init__()
        from transformers import AutoModel
        self.b = AutoModel.from_pretrained(repo)
        self.head = nn.Linear(self.b.config.hidden_size, 2)
    def forward(self, x):
        return self.head(self.b(pixel_values=x).last_hidden_state[:, 0])

FT_SPECS = {
    "vit_b16": ("ViT-B/16 ImageNet fine-tuned",
                lambda: timm.create_model("vit_base_patch16_224", pretrained=True, num_classes=2),
                ((.5, .5, .5), (.5, .5, .5)), CFG["FT_LR"]),
    "phikon": ("Phikon fine-tuned (full)", lambda: HFBackbone("owkin/phikon"), IMN, 1.5e-5),
}

def finetune(build, norm, lr, seed=CFG["SEED"]):
    seed_all(seed)
    m = build().to(DEVICE)
    mean = torch.tensor(norm[0], device=DEVICE).view(1, 3, 1, 1)
    std = torch.tensor(norm[1], device=DEVICE).view(1, 3, 1, 1)
    prep = lambda xb: (xb.to(DEVICE).float() / 255 - mean) / std
    bs = CFG["FT_BS"]
    head = [p for n, p in m.named_parameters() if n.startswith("head")]
    body = [p for n, p in m.named_parameters() if not n.startswith("head")]
    opt = torch.optim.AdamW([dict(params=body, lr=lr), dict(params=head, lr=lr * 20)],
                            weight_decay=0.05)
    steps = CFG["FT_EPOCHS"] * math.ceil(len(TR) / bs)
    sched = torch.optim.lr_scheduler.OneCycleLR(opt, max_lr=[lr, lr * 20],
                                                total_steps=steps, pct_start=0.1)
    scaler = torch.amp.GradScaler("cuda", enabled=AMP)

    @torch.no_grad()
    def predict(Xs):
        m.eval(); o = []
        for i in range(0, len(Xs), 64):
            with torch.autocast("cuda", dtype=torch.float16, enabled=AMP):
                o.append(m(prep(Xs[i:i + 64])).float().softmax(1).cpu())
        return torch.cat(o).numpy()

    best_acc, best_state = -1, None
    for ep in range(CFG["FT_EPOCHS"]):
        m.train(); perm = np.random.permutation(TR)
        for i in range(0, len(perm), bs):
            idx = torch.as_tensor(perm[i:i + bs])
            xb, yb = X224[idx], torch.from_numpy(Y[idx.numpy()]).to(DEVICE)
            if random.random() < .5: xb = xb.flip(3)
            if random.random() < .5: xb = xb.flip(2)
            xb = torch.rot90(xb, random.randint(0, 3), (2, 3))
            with torch.autocast("cuda", dtype=torch.float16, enabled=AMP):
                loss = F.cross_entropy(m(prep(xb)).float(), yb)
            opt.zero_grad(set_to_none=True)
            scaler.scale(loss).backward(); scaler.step(opt); scaler.update(); sched.step()
        pv = predict(X224[VA]); acc = (pv.argmax(1) == Y[VA]).mean()
        print(f"  epoch {ep+1}/{CFG['FT_EPOCHS']}  loss={loss.item():.4f}  val_acc={acc*100:.2f}")
        if acc > best_acc:
            best_acc = acc
            best_state = {k: v.detach().clone() for k, v in m.state_dict().items()}
    m.load_state_dict(best_state)
    out = (predict(X224[VA]), predict(X224[TE]), predict(perturb(X224[TE])))
    del m, opt; torch.cuda.empty_cache()
    return out

for key in CFG["FINETUNE"]:
    label, build, norm, lr = FT_SPECS[key]
    print(f"\n=== {label} ===")
    try:
        p_va, p_te, p_tp = finetune(build, norm, lr)
        PROBS[key] = p_te
        extra[key] = add_row(RES, label, p_va, p_te, p_tp)
    except Exception as e:
        print(f"  failed: {type(e).__name__}: {str(e)[:200]}")
    torch.cuda.empty_cache()

# %% ------------------- 6. CONCEPT BOTTLENECK (PLIP) ------------------
MALIGNANT = ["keratin pearls", "nuclear pleomorphism", "hyperchromatic nuclei",
             "abnormal mitotic figures", "loss of cellular polarity",
             "invasion of tumor cells into the stroma",
             "irregular nests of atypical squamous cells",
             "high nuclear to cytoplasmic ratio", "disorganized epithelium"]
NORMAL = ["regular stratified squamous epithelium",
          "orderly basal cell layer with maturation", "uniform small nuclei",
          "intact basement membrane", "normal keratinized surface layer",
          "connective tissue stroma without tumor"]
CONCEPTS = MALIGNANT + NORMAL
GROUP_RES = {}

cb_name = CFG["CONCEPT_MODEL"]
try:
    if cb_name in FEATS and "model" in PLIP:
        print("\n=== concept bottleneck (PLIP) ===")
        with torch.no_grad():
            tk = PLIP["proc"](text=[f"a histopathology image showing {c}" for c in CONCEPTS],
                              return_tensors="pt", padding=True).to(DEVICE)
            mm = PLIP["model"]
            te_ = mm.text_projection(mm.text_model(**tk).pooler_output).float()
            Tn = F.normalize(te_, dim=-1).cpu().numpy()

        def concept_scores(Fimg):
            Fn_ = Fimg / np.linalg.norm(Fimg, axis=1, keepdims=True)
            return 100 * Fn_ @ Tn.T

        S_all, S_tp = concept_scores(FEATS[cb_name]), concept_scores(FEATS_P[cb_name])
        sc = StandardScaler().fit(S_all[TR]); Z = lambda S: sc.transform(S)
        cbm = LogisticRegressionCV(Cs=[1e-2, 1e-1, 1, 10], cv=3, max_iter=3000,
                                   scoring="neg_log_loss").fit(Z(S_all[TR]), Y[TR])
        p_va, p_te, p_tp = (cbm.predict_proba(Z(S_all[VA])), cbm.predict_proba(Z(S_all[TE])),
                            cbm.predict_proba(Z(S_tp)))
        PROBS["concept_bottleneck"] = p_te
        extra["concept_bottleneck"] = add_row(RES, "PLIP concept bottleneck (15 concepts)",
                                              p_va, p_te, p_tp)
        Zte = Z(S_all[TE]); base_p = cbm.predict_proba(Zte)[:, 1]; rows = []
        for j, c in enumerate(CONCEPTS):
            z0 = Zte.copy(); z0[:, j] = 0
            z2 = Zte.copy(); z2[:, j] += 2
            p0, p2 = cbm.predict_proba(z0)[:, 1], cbm.predict_proba(z2)[:, 1]
            rows.append(dict(concept=c, type="malignant" if c in MALIGNANT else "normal",
                             coef=cbm.coef_[0, j],
                             acc_if_removed=((p0 > .5) == Y[TE]).mean() * 100,
                             mean_dP_plus2sd=(p2 - base_p).mean()))
        CB = pd.DataFrame(rows).sort_values("coef", ascending=False)
        print(CB.round(3).to_string(index=False))
        CB.to_csv(f"{CFG['OUT']}/concept_interventions.csv", index=False)
        plt.figure(figsize=(7, 5))
        plt.barh(CB.concept[::-1], CB.coef[::-1],
                 color=["tab:red" if t == "malignant" else "tab:blue" for t in CB.type[::-1]])
        plt.axvline(0, c="k", lw=.5); plt.xlabel("logistic coefficient (>0 pushes to OSCC)")
        plt.title("Concept bottleneck: what drives the prediction"); plt.tight_layout()
        plt.savefig(f"{CFG['OUT']}/concept_coefficients.png", dpi=150); plt.show()
    else:
        print("\nconcept bottleneck skipped (PLIP not loaded)")
except Exception as e:
    print(f"\nconcept bottleneck failed: {type(e).__name__}: {str(e)[:200]}")

# %% ------------- 7. LEAKAGE AUDIT + GROUP-AWARE CV -------------------
print("\n=== leakage audit ===")
Fg = F_GROUP if F_GROUP is not None else (FEATS[next(iter(FEATS))] if FEATS else None)
try:
    if Fg is not None:
        Fn = Fg / np.linalg.norm(Fg, axis=1, keepdims=True)
        trn = Fn[TR]
        plt.figure(figsize=(7, 4))
        for nm, idx in (("test->train", TE), ("val->train", VA)):
            mx = (Fn[idx] @ trn.T).max(1)
            print(f"  {nm}: median={np.median(mx):.3f}  "
                  + "  ".join(f">{t}: {(mx > t).mean()*100:.1f}%" for t in (.90, .95, .98)))
            plt.hist(mx, bins=40, alpha=.6, label=nm)
        S = trn @ trn.T; np.fill_diagonal(S, -1)
        print(f"  train->train (excl. self): median={np.median(S.max(1)):.3f}")
        plt.xlabel("max cosine similarity to any TRAIN image"); plt.legend()
        plt.title("Near-duplicate audit (D4-averaged embeddings)"); plt.tight_layout()
        plt.savefig(f"{CFG['OUT']}/leakage_hist.png", dpi=150); plt.show()

        print("  cluster counts at different distance thresholds:")
        for t in (0.05, 0.10, 0.15, 0.20, 0.30):
            k = AgglomerativeClustering(n_clusters=None, distance_threshold=t,
                                        metric="cosine", linkage="average").fit(Fn).n_clusters_
            print(f"    thr={t:.2f}: {k} clusters")
        groups = AgglomerativeClustering(n_clusters=None, distance_threshold=CFG["GROUP_DIST_THRESH"],
                                         metric="cosine", linkage="average").fit_predict(Fn)
        print(f"  using thr={CFG['GROUP_DIST_THRESH']}: {groups.max()+1} groups")
        pd.DataFrame(dict(path=PATHS, label=Y, orig_split=SPLIT, group=groups)).to_csv(
            f"{CFG['OUT']}/group_assignments.csv", index=False)

        def cv_acc(Fm, splitter, **kw):
            a = []
            for tr, te in splitter.split(Fm, Y, **kw):
                c = make_pipeline(StandardScaler(), LogisticRegression(C=0.1, max_iter=2000))
                a.append(c.fit(Fm[tr], Y[tr]).score(Fm[te], Y[te]) * 100)
            return np.mean(a), np.std(a)

        skf = StratifiedKFold(CFG["CV_FOLDS"], shuffle=True, random_state=CFG["SEED"])
        gkf = GroupKFold(CFG["CV_FOLDS"])
        for name, Fm in FEATS.items():
            r = cv_acc(Fm, skf); g = cv_acc(Fm, gkf, groups=groups)
            GROUP_RES[name] = dict(random_cv=f"{r[0]:.2f}±{r[1]:.2f}", group_cv=f"{g[0]:.2f}±{g[1]:.2f}",
                                   inflation=float(r[0] - g[0]))
            print(f"  {name:10s} random-CV {r[0]:.2f}±{r[1]:.2f} | group-aware CV {g[0]:.2f}±{g[1]:.2f}"
                  f" | inflation {r[0]-g[0]:+.2f} pts")
    else:
        print("  no features available for audit")
except Exception as e:
    print(f"  audit failed: {type(e).__name__}: {str(e)[:200]}")

# %% ------------------------ 8. FINAL TABLE ---------------------------
df = pd.DataFrame(RES)
for i, r in df.iterrows():
    key = r["model"].split(" + ")[0]
    if key in GROUP_RES:
        df.loc[i, "random_cv"] = GROUP_RES[key]["random_cv"]
        df.loc[i, "group_cv"] = GROUP_RES[key]["group_cv"]
df["beats_paper"] = (df.acc > BASE["acc"]) & (df.prec > BASE["prec"]) & (df.sens > BASE["sens"])
cols = ["model", "acc", "ci_lo", "ci_hi", "prec", "sens", "spec", "f1", "auc", "ece",
        "acc_stain", "cp_coverage", "cp_avg_set_size", "cp_acc_on_singleton",
        "random_cv", "group_cv", "beats_paper"]
cols = [c for c in cols if c in df.columns]
pd.set_option("display.width", 250); pd.set_option("display.max_columns", 50)
print("\n" + "=" * 100)
print(f"TARGET (paper ViT-14): acc>{BASE['acc']}  prec>{BASE['prec']}  sens>{BASE['sens']}")
print("=" * 100)
print(df[cols].round(2).to_string(index=False))
df[cols].round(3).to_csv(f"{CFG['OUT']}/results.csv", index=False)
json.dump(dict(cfg=CFG, base=BASE, n=dict(train=len(TR), val=len(VA), test=len(TE)),
               conformal=extra, group_cv=GROUP_RES),
          open(f"{CFG['OUT']}/results.json", "w"), indent=2, default=str)
print(f"\nSaved results.csv, results.json, leakage_hist.png, concept_* and "
      f"group_assignments.csv -> {CFG['OUT']}")
print(f"NOTE: test set has only {len(TE)} images ({int((Y[TE]==0).sum())} normal); "
      f"1 error = {100/len(TE):.2f} pts. Use the CI columns, not the 2nd decimal.")
```
